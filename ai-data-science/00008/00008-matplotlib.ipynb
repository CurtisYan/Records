```python
import pandas as pd
orders = pd.DataFrame({
    "order_id": ["O101", "O102", "O103", "O104"],
    "user_id": ["U01", "U01", "U02", "U03"],
    "amount": [100, 50, 80, 60],
    "paid": [True, True, True, False],
})
paid_flag = orders["paid"]
paid_orders = orders[paid_flag]
by_user = paid_orders.groupby("user_id")
grouped_amounts = by_user["amount"]
user_amounts = grouped_amounts.sum()
user_counts = by_user.size()
user_summary = pd.DataFrame({
    "paid_order_count": user_counts,
    "paid_amount": user_amounts,
})
```

```python
import matplotlib.pyplot as plt
x = user_summary.index  # U01 U02
y = user_summary["paid_amount"] # 150 80
fig, ax = plt.subplots()
ax.bar(x, y)
ax.set_xlabel("User ID")
ax.set_ylabel("Paid amount (CNY)")
ax.set_title("Paid amount by user")
ax.set_ylim(0, 180)
plt.show()
```

```python
order_amounts = paid_orders["amount"]   # 100 50 80
fig, ax = plt.subplots()
counts, edges, patches = ax.hist(   # 各区间计数、区间边界、画出的柱形对象
    order_amounts, bins=[0, 60, 120], density=False
)
ax.set_xlabel("Order amount (CNY)")
ax.set_ylabel("Order count")
plt.show()
```
