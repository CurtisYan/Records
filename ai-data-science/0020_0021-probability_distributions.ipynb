### 期望与方差
![img](img/image-3.png)

```python
import numpy as np
x = np.array([-10., 0., 30.])
p = np.array([0.25, 0.50, 0.25])
mu = (x * p).sum()
variance = ((x - mu) ** 2 * p).sum()
sd = np.sqrt(variance)
print(mu, variance, sd)
```

### 常用概率分布    
[正态分布](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.norm.html)

[二项分布](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.binom.html)

```python
# 二项分布与正态分布

# 二项分布通常用于计算固定次数的独立重复试验中，某件事情成功多少次的概率
# 正态分布则是一种常见的连续概率分布，图像呈钟形

from scipy.stats import binom, norm
# pmf:Probability Mass Function，概率质量函数
exact_small = binom.pmf(2, n=3, p=0.5)
# cdf：Cumulative Distribution Function，累积分布函数
exact = binom.cdf(60, n=100, p=0.5)

# norm.cdf: 正态分布的累积分布函数
approx = norm.cdf(60.5, loc=50, scale=5)
print(round(exact_small, 4))
print(round(exact, 4), round(approx, 4))
```
